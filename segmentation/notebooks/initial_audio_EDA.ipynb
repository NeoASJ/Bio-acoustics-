```python
import librosa
import numpy as np
```

# Peter's Dwarf Frog (Engystomops petersi)

```python
from IPython import display
display.Image("https://upload.wikimedia.org/wikipedia/commons/7/71/Engystomops_petersi_1.jpg")
```

```python
import simpleaudio as sa
 
# Define audio
wave_object = sa.WaveObject.from_wave_file("LS100158 E. petersi no capturado_Daniel1.wav")
 
# Define object to control play
play_object = wave_object.play()
```

```python
# Stop playing audio
play_object.stop()
```

```python
# Load data
y, sr = librosa.load('LS100158 E. petersi no capturado_Daniel1.wav')
# y = waveform of audio
# sr = sampling rate
```

```python
sr
```

```python
import matplotlib.pyplot as plt

# Construct a subplot grid with 3 rows and 1 column, sharing the x-axis)
fig, ax = plt.subplots(nrows=2, ncols=1, sharex=True)

# On the first subplot, show the original spectrogram
img1 = librosa.display.waveshow(y, sr=sr, axis='s', color='green', ax=ax[0])
ax[0].set(title='Original waveform')
ax[0].set_xlim(1, 1.75)

# On the second subplot, show the mel spectrogram
S = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128)
S_dB = librosa.power_to_db(S, ref=np.max)
img2 = librosa.display.specshow(S_dB, x_axis='s',
                                y_axis='mel', sr=sr,
                                ax=ax[1])
ax[1].set(title='Mel-frequency spectrogram')

# To eliminate redundant axis labels, we'll use "label_outer" on all subplots:
for ax_i in ax:
    ax_i.label_outer()
    
# And we can share colorbars:
fig.colorbar(img2, ax=[ax[0], ax[1]], format='%+2.0f dB')
```
