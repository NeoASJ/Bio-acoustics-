```python
import librosa, librosa.display
import numpy as np
import matplotlib.pyplot as plt
import os
os.chdir('/Users/meldrumapple/Desktop/DL')
```

```python
x, sr = librosa.load('H.lanciformis2.WAV')
onset_frames = librosa.onset.onset_detect(y=x, sr=sr, wait=1, pre_avg=1, post_avg=1, pre_max=1, post_max=1) #
#print(onset_frames) # frame numbers of estimated onsets
onset_times = librosa.frames_to_time(onset_frames) #

plt.figure(figsize=(20, 5))
librosa.display.waveshow(x, sr=sr)
plt.title('Waveplot', fontdict=dict(size=18))
plt.vlines(onset_times, 0,ymax=0.6, color='r', alpha=0.9, linestyle='--', 
label='Onsets') #
plt.xlabel('Time', fontdict=dict(size=15))
plt.ylabel('Amplitude', fontdict=dict(size=15))
plt.show()
```
